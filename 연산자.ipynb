## 1-2. 증감연산자
  
<span style="display:inline-block;postion:left;">  
  
| 연산자   | 명칭    | 기능                              |
| ----- | ----- | ------------------------------- |
| `++a` | 전위 증가 | 먼저 증가시킨 후 해당 식의 값을 사용           |
| `a++` | 후위 증가 | 해당 식의 값은 증가 전 값을 사용하고, 증가 작업 수행 |
| `--a` | 전위 감소 | 먼저 감소시킨 후 해당 식의 값을 사용           |
| `a--` | 후위 감소 | 해당 식의 값은 감소 전 값을 사용하고, 감소 작업 수행 |


</span>

```c++
// 결과는
int a = 3;
printf("1. a = %d\n", ++a);
printf("2. a = %d\n", a++);
printf("3. a = %d\n", a--);
printf("4. a = %d\n", --a);
```

```c++
// 나머지 연산
printf("-13%3 나머지: %d\n", -13%3); // 몫
printf("-13%-3 나머지: %d", -13%-3); // 나머지
```

### 나머지 연산 예시

<div style="display:inline-block;position:left;">

| 연산식       | 몫의 계산                    |  나머지 |
| --------- | ------------------------ | ---: |
| `-7 % -2` | `-7 / -2 = 3` (소수 부분 버림) | `-1` |
| `-7 % 2`  | `-7 / 2 = -3` (소수 부분 버림) | `-1` |
| `7 % -2`  | `7 / -2 = -3` (소수 부분 버림) |  `1` |
| `7 % 2`   | `7 / 2 = 3`              |  `1` |

</div>

* C의 나머지 연산규칙 : a  = (a/b)*b + (a%b)
* C의 나머지 연산은 수학의 양수 모듈 연산과 다름

```c++
/* 다음을 나머지 연산으로 풀어보라
* 3자리 양의 정수를 입력한다. (예: 354  또는 785)
* 입력된 값이 354일때 출력형태:
  백의자리: 3
  십의자리: 5
  일의자리: 4
*/
//프로그램 구현
#include <stdio.h>
int main(){
    int num = 0;
    printf("백의 자리 양의 정수값을 입력하시오");
    scanf("%d",&num);
    //여기서부터 구현
}
```

## 2. 대입연산자

<span style="position:left;display:inline-block;">

| 연산자   | 의미           | 사용 예      | 동등한 표현       |
| ----- | ------------ | --------- | ------------ |
| `=`   | 대입           | `a = 10`  | 값을 대입        |

| `%=`  | 나머지를 대입      | `a %= 3`  | `a = a % 3`  |

| `\|=` | 비트 OR 후 대입   | `a \|= b` | `a = a \| b` |
| `^=`  | 비트 XOR 후 대입  | `a ^= b`  | `a = a ^ b`  |
| `<<=` | 왼쪽 시프트 후 대입  | `a <<= 1` | `a = a << 1` |
| `>>=` | 오른쪽 시프트 후 대입 | `a >>= 1` | `a = a >> 1` |


</span>

## 4. 논리연산자

<span style="display:inline-block;position:left;">

| 연산자    | 명칭     | 의미           | 사용 예               |
| ------ | ------ | ------------ | ------------------ |
| `&&`   | 논리 AND | 두 조건이 모두 참   | `a > 0 && b > 0`   |
| `\|\|` | 논리 OR  | 하나 이상의 조건이 참 | `a > 0 \|\| b > 0` |
| `!`    | 논리 NOT | 참과 거짓을 반전    | `!(a > 0)`         |


</span>

### 논리곱
 

```c++
// 논리곱
#include <stdio.h>
int main(){
    0 && printf("0");
    1 && printf("1");
}
```

### 논리합


```c++
// 논리합
#include <stdio.h>
int main(){
    0 || printf("0");
    1 || printf("1");
}
```

### 문제


```c++
// 아래 코드의 while문 조건을 드모르간 법칙에 따라 재구성하라
#include <stdio.h>
int main(){
    int no;
    printf("드모르간 법칙을 적용하여 while 루프의 조건을 변경하시오.\n");
    do {
        printf("2자리 양수 입력: ");
        scanf("%d",&no);
    }while(no < 10 || no > 99);
    printf("입력값: %d\n",no);
    return 0;
}
```

## 5. 비트연산자

<span style="display:inline-block;position:left;">

| 연산자  | 명칭      | 기능                |
| ---- | ------- | ----------------- |
| `&`  | 비트 AND  | 두 비트가 모두 1이면 1    |
| `\|` | 비트 OR   | 두 비트 중 하나라도 1이면 1 |
| `^`  | 비트 XOR  | 두 비트가 서로 다르면 1    |
| `~`  | 비트 NOT  | 각 비트를 반전          |
| `<<` | 왼쪽 시프트  | 비트를 왼쪽으로 이동       |
| `>>` | 오른쪽 시프트 | 비트를 오른쪽으로 이동      |


</span>

```c++
// 비트연산
printf("%hhb & %hhb = %hhb\n", 0b10101010, 0b11010111, 0b10101010&0b11010111);
printf("%hhb | %hhb = %hhb\n", 0b10101010, 0b01010110, 0b10101010|0b01010110);
printf("%hhb ^ %hhb = %hhb\n", 0b10101010, 0b00101011, 0b10101010^0b00101011);
printf("~0%hhb = %hhb\n", 0b01111110, ~0b01111110);
printf("%hhb<<3 = 0b%hhb = %d\n", 0b00000001, 0b00000001<<3, 0b00000001<<3);
printf("%hhb>>4 = 0b%hhb = %d\n", 0b00010000, 0b00010000>>4, 0b00010000>>4);
```

## 7. 기타연산자

<span style="display:inline-block;position:left;">

| 연산자      | 명칭        | 기능                        | 사용 예              |
| -------- | --------- | ------------------------- | ----------------- |
| `sizeof` | 크기 연산자    | 자료형 또는 객체의 크기를 바이트 단위로 확인 | `sizeof(int)`     |
| `&`      | 주소 연산자    | 객체의 메모리 주소를 구함            | `&a`              |
| `*`      | 역참조 연산자   | 포인터가 가리키는 객체에 접근          | `*ptr`            |
| `,`      | 쉼표 연산자    | 왼쪽 식을 평가한 뒤 오른쪽 식의 결과를 사용 | `(a = 1, a + 2)`  |
| `.`      | 구조체 멤버 접근 | 구조체 객체의 멤버에 접근            | `student.age`     |
| `->`     | 포인터 멤버 접근 | 구조체 포인터를 통해 멤버에 접근        | `ptr->age`        |
| `[]`     | 배열 첨자     | 배열 원소에 접근                 | `arr[0]`          |
| `()`     | 함수 호출     | 함수를 호출                    | `printf("Hello")` |
| `(자료형)`  | 형변환       | 값을 지정한 자료형으로 변환           | `(double)a`       |


</span>

```c++
/* 문제 : 윤년 분별기 만들기
// 윤년이란 4의 배수이면서 100의 배수가 아니거나, 400의 배수인 해
// 입력폼: 년도를 입력하시오(예: 1992 )
// 출력폼: 1992은 윤년이 아닙니다. 
*/
#include <stdio.h>
int main(){
    int year;
    printf("년도를 입력하시오(예: 1992)");
    scanf("%d", &year);
    if((year%4==0 && year%100 != 0) || (year%400==0)) 
        printf("%d는 윤년입니다.", year);
    else
        printf("%d는 평년입니다.", year);
}
    
```
